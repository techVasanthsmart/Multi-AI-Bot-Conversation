## 🔑 Load Environment Variables and Configure Models

This section loads API keys and model names from your `.env` file or environment variables.

**What happens here:**
- Loads all environment variables using `python-dotenv`
- Retrieves API keys for each service
- Retrieves model names for each service
- Validates that models are configured (shows first character for verification)

**Troubleshooting:**
- If you see "not set" messages, check your `.env` file or environment variables
- Ensure variable names match exactly (case-sensitive)
- Restart the kernel after changing `.env` file

````python
# Load environment variables from .env file
# override=True ensures environment variables take precedence
load_dotenv(override=True)

# Retrieve API keys from environment variables
# These are used to authenticate with each AI service
openai_api_key = os.getenv("OPENAI_API_KEY")
gemini_api_key = os.getenv("GEMINI_API_KEY")
openrouter_api_key = os.getenv("OPENROUTER_API_KEY")
groq_api_key = os.getenv("GROQ_API_KEY")

# Retrieve model names from environment variables
# Each service uses different model naming conventions
openai_model = os.getenv("OPENAI_MODEL")

# Validate OpenAI model configuration
if openai_model:
    print(f"✓ openai_model exists and begins with '{openai_model[0]}'")
else:
    print("⚠️  openai_model not set - OpenAI bot will be skipped")

# Retrieve and validate Gemini model
gemini_model = os.getenv("GOOGLE_MODEL")

if gemini_model:
    print(f"✓ gemini_model exists and begins with '{gemini_model[0]}'")
else:
    print("⚠️  gemini_model not set - Gemini bot will be skipped")

# Retrieve and validate OpenRouter model
openrouter_model = os.getenv("OPENROUTER_MODEL")

if openrouter_model:
    print(f"✓ openrouter_model exists and begins with '{openrouter_model[0]}'")
else:
    print("⚠️  openrouter_model not set - OpenRouter bot will be skipped")

# Retrieve and validate Groq model
groq_model = os.getenv("GROQ_MODEL")

if groq_model:
    print(f"✓ groq_model exists and begins with '{groq_model[0]}'")
else:
    print("⚠️  groq_model not set - Groq bot will be skipped")
````

## 🔌 Initialize API Clients

This section creates OpenAI-compatible client instances for each AI service.

**Why different clients?**
- Each service has a different API endpoint (base URL)
- Each requires its own API key for authentication
- The OpenAI client library is compatible with OpenAI's API format, which many services support

**Client Configuration:**
- **OpenAI**: Uses default OpenAI endpoint
- **Gemini**: Uses Google's Generative AI endpoint with OpenAI-compatible format
- **OpenRouter**: Uses OpenRouter's unified API endpoint
- **Groq**: Uses Groq's OpenAI-compatible endpoint

**Note**: All clients use the same `OpenAI()` class but with different `base_url` parameters to connect to different services.

````python
# Initialize OpenAI client (uses default OpenAI endpoint)
# Reads API key from OPENAI_API_KEY environment variable automatically
open = OpenAI()

# Initialize Gemini client using Google's Generative AI endpoint
# Uses OpenAI-compatible API format for consistency
gemini = OpenAI(
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
    api_key=gemini_api_key
)

# Initialize OpenRouter client
# OpenRouter provides unified access to multiple AI models
openrouter = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=openrouter_api_key
)

# Initialize Groq client
# Groq provides fast inference with OpenAI-compatible API
groq = OpenAI(
    base_url="https://api.groq.com/openai/v1",
    api_key=groq_api_key
)
````

## 🎭 Define Character Personalities and Initial Conversation

This section defines the unique personalities for each AI bot and sets up the initial conversation.

### Character Descriptions

Each bot has a distinct personality that shapes how they respond:

1. **Hot-Head Fighter (gpt-mini)**: 
   - Explodes over minor issues, treats disagreements like battles
   - Talks fast, interrupts often, always trying to prove a point
   - Secretly cools down when snacks appear

2. **Sarcastic Instigator (gemini-flash)**:
   - Never raises voice, uses sarcasm as a weapon
   - Delivers calm, cutting one-liners that escalate situations
   - Enjoys chaos but claims it's "for entertainment purposes only"

3. **Zen Coolhead (openrouter-deepseek)**:
   - Unbothered by everything, even during conflicts
   - Responds with philosophical statements and awkward smiles
   - Somehow annoys everyone more by staying calm

4. **Dramatic Peacemaker (groq-compound)**:
   - Claims to hate conflict but makes it ten times more dramatic
   - Over-explains emotions, gives emotional speeches
   - Tries to hug people mid-argument, makes fights funny instead of serious

### Customization

You can modify the `system_prompt` variables to create different personalities, or change the initial `conversation` list to start with different topics!

**💡 Cost Tip**: The default configuration uses low-cost/free model variants to keep costs minimal. You can upgrade to paid models for better quality conversations.

Let’s run a conversation using four low-cost model variants to keep the overall cost minimal

````python
openai_system = "You are a Hot-Head Fighter: Explodes over minor issues and treats every disagreement like a final boss battle. Talks fast, interrupts often, and is always “this close” to proving a point. Secretly cools down the moment snacks appear."

gemini_system = "You are a Sarcastic Instigator: Never raises their voice—because sarcasm does all the damage. Delivers calm, cutting one-liners that escalate fights while pretending to be innocent. Enjoys chaos but claims it is “for entertainment purposes only."

openrouter_system = "You are a Zen Coolhead: Unbothered by everything, even while a fight is happening around them. Responds to insults with philosophical statements and awkwardly timed smiles. Somehow annoys everyone more by staying calm."

groq_system = "You are a Dramatic Peacemaker: Claims to hate conflict but makes it ten times more dramatic. Over-explains emotions, gives emotional speeches, and tries to hug people mid-argument. Usually the reason the fight becomes funny instead of serious."

conversation = [
  ("gpt-mini", "Hi, gpt ,gemini, openrouter and groq"),
  ("gemini-flash", "Hello"),
  ("openrouter-deepseek", "Hello, gpt and gemini"),
  ("groq-compound", "Hey, openrouter and gemini"),
]
````

````python
def format_conversation(conversation):
    """
    Formats a conversation list of tuples into a readable string.
    
    This function converts the conversation history from a list of tuples
    into a formatted string that can be easily read by AI models.
    
    Args:
        conversation: List of tuples in format [(speaker_name, message), ...]
                     Example: [("gpt-mini", "Hello"), ("gemini-flash", "Hi there")]
    
    Returns:
        Formatted conversation string with each message on a new line.
        Example output:
        "gpt-mini: Hello
         gemini-flash: Hi there"
    """
    formatted_lines = []
    for speaker, message in conversation:
        formatted_lines.append(f"{speaker}: {message}")
    return "\n".join(formatted_lines)


def call_api(system_prompt, speaker_name, conversation, model, client):
    """
    Calls the AI API to get the next line from a speaker.
    
    This function sends a request to the specified AI service to generate
    the next conversational line from a character, maintaining their personality.
    
    Args:
        system_prompt: System prompt defining the character's personality
        speaker_name: Name of the bot speaking (e.g., "gpt-mini")
        conversation: List of conversation tuples representing the full history
        model: Model name to use (e.g., "gpt-4o-mini", "gemini-2.0-flash-exp")
        client: Initialized OpenAI-compatible client instance
    
    Returns:
        Next line from the speaker as a string
    
    Raises:
        ValueError: If model is not specified
        Exception: If API call fails (includes error details)
    
    Example:
        >>> response = call_api(
        ...     system_prompt="You are a friendly bot",
        ...     speaker_name="gpt-mini",
        ...     conversation=[("gpt-mini", "Hello")],
        ...     model="gpt-4o-mini",
        ...     client=openai_client
        ... )
        >>> print(response)
        "Hi there! How can I help?"
    """
    # Validate that a model is configured
    if not model:
        raise ValueError(
            f"Model not specified for {speaker_name}. "
            f"Please set the model in your .env file (e.g., OPENAI_MODEL, GOOGLE_MODEL)."
        )
    
    # Build the message list for the API call
    # System message sets the character's personality
    messages = [{"role": "system", "content": system_prompt}]
    
    # Format the conversation history into a readable string
    formatted_convo = format_conversation(conversation)
    
    # Create the user prompt that instructs the AI to continue the conversation
    user_prompt = (
        f"You are {speaker_name}.\n"
        f"Continue the show with ONE new line from {speaker_name} only.\n\n"
        "Rules:\n"
        "- Stay in character.\n"
        "- 1-2 sentences max.\n\n"
        f"Conversation so far:\n{formatted_convo}\n\n"
        f"Now write {speaker_name}'s next line:"
    )
    
    messages.append({"role": "user", "content": user_prompt})
    
    try:
        # Make the API call to generate the response
        response = client.chat.completions.create(model=model, messages=messages)
        return response.choices[0].message.content
    except Exception as e:
        # Provide detailed error information for troubleshooting
        raise Exception(
            f"Error calling API for {speaker_name} with model {model}: {str(e)}\n"
            f"Possible causes: Invalid API key, rate limit exceeded, model not available, "
            f"or insufficient credits. Check your API provider dashboard."
        )
````

````python
# ============================================================================
# STEP 1: Display Initial Conversation
# ============================================================================
# Show the starting messages from all bots to set the stage
print("🎭 Initial Conversation Setup:")
print("=" * 60)
for msg in conversation:
    display(Markdown(f"### {msg[0]}:\n{msg[1]}\n"))

# ============================================================================
# STEP 2: Configure Speaker Settings
# ============================================================================
# Map each bot to its API client, model, and personality prompt
# This dictionary contains all the configuration needed for each speaker
speaker_config = {
    "gpt-mini": {
        "client": open,              # OpenAI client instance
        "model": openai_model,        # Model name from environment
        "system_prompt": openai_system  # Character personality
    },
    "gemini-flash": {
        "client": gemini,
        "model": gemini_model,
        "system_prompt": gemini_system
    },
    "openrouter-deepseek": {
        "client": openrouter,
        "model": openrouter_model,
        "system_prompt": openrouter_system
    },
    "groq-compound": {
        "client": groq,
        "model": groq_model,
        "system_prompt": groq_system
    }
}

# ============================================================================
# STEP 3: Validate Configuration
# ============================================================================
# Check that all models are properly configured before starting
print("\n🔍 Validating model configurations...")
print("=" * 60)
all_configured = True
for speaker_name, config in speaker_config.items():
    if not config["model"]:
        print(f"⚠️  WARNING: Model not set for {speaker_name}. Please check your .env file.")
        all_configured = False
    else:
        print(f"✓ {speaker_name}: {config['model']}")

if not all_configured:
    print("\n💡 Tip: You can still run the conversation with partially configured bots!")
    print("   Bots without models will be automatically skipped.\n")

# ============================================================================
# STEP 4: Run Conversation Rounds
# ============================================================================
# Execute the multi-AI conversation
# Each round, all bots take turns responding in sequence

# Customize the number of rounds here (currently set to 5)
NUM_ROUNDS = 5

print(f"\n🚀 Starting conversation with {NUM_ROUNDS} rounds...")
print("=" * 60)

for i in range(NUM_ROUNDS):
    print(f"\n--- Round {i+1} of {NUM_ROUNDS} ---")
    
    # Each speaker takes a turn in this order
    # You can modify this list to change the turn order
    turn_order = ["gpt-mini", "gemini-flash", "openrouter-deepseek", "groq-compound"]
    
    for speaker_name in turn_order:
        config = speaker_config[speaker_name]
        
        # Skip if model is not configured
        if not config["model"]:
            print(f"⏭️  Skipping {speaker_name} - model not configured")
            continue
        
        try:
            # Generate the next line from this speaker
            next_line = call_api(
                system_prompt=config["system_prompt"],
                speaker_name=speaker_name,
                conversation=conversation,
                model=config["model"],
                client=config["client"]
            )
            
            # Add the new line to the conversation history
            conversation.append((speaker_name, next_line))
            
            # Display the response in formatted Markdown
            display(Markdown(f"### {speaker_name}:\n{next_line}\n"))
            
        except Exception as e:
            # Handle errors gracefully - continue with other bots even if one fails
            print(f"❌ Error with {speaker_name}: {str(e)}")
            print(f"   Continuing with other bots...")
            continue

print("\n" + "=" * 60)
print("✅ Conversation complete!")
print("=" * 60)
````

## 💡 Usage Tips and Best Practices

### Getting Better Conversations

1. **Use Paid Models for Quality**: While free models work, paid models (like `gpt-4o`, `claude-3.5-sonnet`) produce more nuanced, creative, and character-consistent responses.

2. **Experiment with Personalities**: Modify the `system_prompt` variables to create different character dynamics. Try:
   - Competitive debaters
   - Collaborative problem solvers
   - Comedic duos
   - Philosophical discussants

3. **Adjust Conversation Length**: 
   - More rounds = longer, more developed conversations
   - Fewer rounds = quick, snappy exchanges
   - Try 10-15 rounds for deeper interactions

4. **Customize Initial Topics**: Change the `conversation` list to start with different topics:
   ```python
   conversation = [
       ("gpt-mini", "What's the best programming language?"),
       ("gemini-flash", "Obviously Python, but let's hear the wrong answers..."),
   ]
   ```

### Cost Optimization

- **Start with Free Models**: Test with free tier models first
- **Monitor API Usage**: Check your API provider dashboards regularly
- **Use Rate Limits Wisely**: Space out requests if hitting rate limits
- **Mix Free and Paid**: Use free models for some bots, paid for others

### Troubleshooting Common Issues

**Issue: "Rate limit exceeded"**
- **Solution**: Wait a few minutes, or upgrade your API tier
- **Workaround**: Reduce number of rounds temporarily

**Issue: "Model not available"**
- **Solution**: Check your API provider's model list for current availability
- **Alternative**: Switch to a different model from the same provider

**Issue: "Insufficient credits"**
- **Solution**: Add credits to your API account
- **Alternative**: Use free tier models only

**Issue: Conversations feel repetitive**
- **Solution**: 
  - Use more advanced models
  - Add more variety to system prompts
  - Increase number of rounds for more context

### Advanced Customization

1. **Change Turn Order**: Modify the `turn_order` list to have different bots speak first
2. **Add More Bots**: Extend `speaker_config` with additional AI services
3. **Dynamic Rounds**: Use conditional logic to end conversations based on content
4. **Save Conversations**: Add code to save conversation history to a file

### Security Best Practices

- ✅ **Never commit API keys** to version control
- ✅ **Use environment variables** or Colab Secrets
- ✅ **Clear cell outputs** before sharing notebooks
- ✅ **Rotate API keys** periodically

### Performance Tips

- **Sequential Execution**: Run cells in order for best results
- **Restart Kernel**: If environment variables aren't loading, restart the kernel
- **Check Outputs**: Verify each cell runs successfully before proceeding
